```python
import torch
```

```python
names = [l.strip() for l in open('names.txt').readlines()]
```

```python
chars = list(set([c for name in names for c in name]))
chars.sort()
chars = ['.'] + chars
```

```python
stoi = {c:i for i, c in enumerate(chars)}
itos = {i:c for i, c in enumerate(chars)}
```

```python
N = torch.zeros(27, 27)
```

```python
for name in names:
    for c1, c2 in zip('.' + name, name + '.'):
        x1 = stoi[c1]
        x2 = stoi[c2]
        N[x1, x2] += 1
```

```python
P = N.float()
P /= N.sum(1, keepdim=True)
```

```python
for _ in range(20):
    w = ''
    ix = 0
    while True:
        ix = torch.multinomial(P[ix], num_samples=1).item()
        if ix == 0:
            break
        w += itos[ix]
    print(w)
```

```python
n = 0
loss = 0.0
for name in ['johnqx']:
    name = '.' + name + '.'
    for c1, c2 in zip(name, name[1:]):
        ix1, ix2 = stoi[c1], stoi[c2]
        prob = P[ix1, ix2]
        loss += torch.log(prob)
        n += 1
        print(f'{c1}{c2}: {prob} {torch.log(prob):.4}')

print(-1 * loss / n)
```

```python
xs = []
ys = []
for name in names:
    for c1, c2 in zip('.' + name, name + '.'):
        x1 = stoi[c1]
        x2 = stoi[c2]
        xs.append(x1)
        ys.append(x2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)
```

```python
import torch.nn.functional as F
xenc = F.one_hot(xs, 27).float()
```

```python
W = torch.randn((27, 27), requires_grad=True)
```

```python
for _ in range(500):
    logits = xenc @ W
    P = logits.exp()
    P = P / P.sum(1, keepdim=True)


    loss = -P[torch.arange(0, len(ys)), ys].log().mean()

    print(loss)

    W.grad = None
    loss.backward()
    W.data += -1 * W.grad * 50
```

```python
for _ in range(10):
    x = 0
    name = ''
    while True:
        enc = F.one_hot(torch.tensor(x), num_classes=27).float()
        p = (enc @ W).exp()
        p /= p.sum()
        x = p.multinomial(1).item()
        if x == 0:
            break
        name += itos[x]
    print(name)
```
